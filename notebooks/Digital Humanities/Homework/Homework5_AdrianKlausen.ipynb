```python
import pandas as pd
df = pd.read_csv("HomeworkDATA.csv") #This dataset contains the 13 largest Estonian theatres by number of visitors. (From my week 4 homework submission.) 
df 
```

-----------------------------------------------------------------------------------------------------------------------------------

```python
import matplotlib.pyplot as plt
df.plot(kind="pie", y="TOTAL VISITORS", labels=df["Theatre"], legend=False, autopct="%1.f%%")
plt.title("Largest Estonian theatres by number of visitors in 2025")
plt.ylabel("")
plt.show()
```

This chart lets us examine the visitation numbers in proportion to eachother. As we can see, Vanemuine was the largest theatre of 2025 by the number of visitors, with Eesti Draamateater close behind.

-----------------------------------------------------------------------------------------------------------------------------------

```python
df.plot(kind="bar", x="Theatre", y=["in-house visitors", "off-site visitors"], stacked=True)
plt.title("In-house and off-site visitors")
plt.ylabel("Number of visitors")
plt.xlabel("Theatre")
plt.legend
plt.show()
```

This chart shows the number of in-house and off-site visitors for each theatre. A stacked bar chart is useful, because with it, we can see both the total number of vistors, and what that total is comprised of.
Endla Teater is the largest one on the chart to have 50% or more of its viewers be off-site. Notably, Teater Sõber only has off-site visitors (and off-site perfomances).

-----------------------------------------------------------------------------------------------------------------------------------

```python
df["VISITORS PER PERFORMANCE"] = (df["TOTAL VISITORS"] / df["TOTAL PERFORMANCES"])
df.plot(kind="bar", x="Theatre", y="VISITORS PER PERFORMANCE", legend=False, color="#1dcf32")
plt.title("Visitors per Performance")
plt.ylabel("Visitors per performance")
plt.xlabel("Theatre")
plt.show()
```

This chart shows each theatre's number of visitors per peformance (dividing the number of visitors by the number of performances). Here we can observe some differences between larger and smaller theatres: the 4th largest theatre by total visitor count, has a smaller performance/visiter ratio than several theatres behind it.
