```python
df = pd.read_csv("competition.csv")
df.describe() #show only rows containing numbers + add statistics
```

```python
import matplotlib.pyplot as plt #matplotlib creates charts and visualisations
df.plot(kind="bar", x="Name", y="Years_of_practice", legend=False)
plt.title("Athletes' years of practice")
plt.ylabel("years of practice")
plt.show()
```

```python
df.plot(kind="bar", x="Name", y="Age", legend=False)
plt.title("Athletes' ages")
plt.ylabel("Age")
plt.show()
```

```python
df.plot(kind="pie", y="Age", labels=df["Name"], legend=False, autopct="%1.1f%%")
plt.title("Athletes' ages")
```

```python
plt.scatter(df["Age"], df["Years_of_practice"], s=df["Years_of_practice"]*60, alpha=0.6) 
#s=df["Years_of_practice"]*60 larger year=bigger bubble, alpha=0.6 makes the bubbles transparent
plt.ylabel("Years of practice")
plt.xlabel("Age")
```

```python
import seaborn as sns #for cleaner box plots
sns.boxplot(data=df, y="Age")
plt.show()
```

```python
plt.Figure(figsize=(8,5))
df[["Age","Years_of_practice"]].boxplot()
plt.show()
```

```python
avg_age= df.groupby("Sport")["Age"].mean() #Getting the average age
avg_age.plot(kind="bar")
plt.show()
```

```python
avg_yr = df.groupby("Sport")["Years_of_practice"].mean()
avg_yr.plot(kind="bar")
plt.title("Average years of practice by sport")
plt.show()
```
