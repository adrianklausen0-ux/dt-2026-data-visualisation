```python
import pandas as pd
df = pd.read_csv("competencies.csv")
df
```

```python
import matplotlib.pyplot as plt
avg_st= df.groupby("Student")["Hours_Studied"].mean()
avg_st.plot(kind="bar", color="#692151")
plt.ylabel("Hours studied") #Making a chart to show the average hours studied.
```

```python
plt.scatter(df["Hours_Studied"], df["Grade"]) 
plt.ylabel("Hours studied")
plt.xlabel("Grade")
```

```python
import seaborn as sns
sns.boxplot(data=df, y="Hours_Studied", x="Grade")
plt.show()
```
